```python
from google import genai
from google.colab import userdata

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

print("Text summarizer connected successfully!")
```

```python
def summarize_text(text, summary_length="short"):
    prompt = f"""
    You are a text summarization assistant.

    Summarize the following text by:
    1. Keeping the main ideas and important facts.
    2. Removing unnecessary details and repetition.
    3. Using clear, simple English.
    4. Not adding information that is not in the original text.
    5. Keeping the summary {summary_length}.

    Original text:
    {text}

    Summary:
    """

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text


print("Text summarizer is ready!")
```

```python
long_text = """
Technology has become an important part of modern education.
Students use computers, smartphones, and online learning
platforms to access educational materials. These tools allow
students to learn from anywhere and at their own pace.
Teachers can use digital resources to explain difficult
concepts and communicate with students. However, excessive
use of technology can cause distractions and reduce face-to-face
interaction. Therefore, technology should be used effectively
to improve the quality of education.
"""

summary = summarize_text(long_text)

print("ORIGINAL TEXT:\n")
print(long_text)

print("\nSUMMARY:\n")
print(summary)
```

```python
while True:
    text = input(
        "\nEnter a long paragraph (or type 'quit' to stop): "
    )

    if text.lower().strip() == "quit":
        print("Text summarizer closed.")
        break

    if not text.strip():
        print("Please enter some text.")
        continue

    try:
        summary = summarize_text(text)
        print("\nSUMMARY:\n")
        print(summary)

    except Exception as error:
        print("Error:", error)
```
